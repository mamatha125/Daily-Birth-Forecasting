```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from collections import Counter

```

```python
file_path = Path("data/Nat2024PublicUS.c20250512.r20250708.txt")

print("File exists:", file_path.exists())

if file_path.exists():
    print("File size:", round(file_path.stat().st_size / (1024**3), 2), "GB")
```

```python
from collections import Counter

monthly_counts = Counter()

total_records = 0
valid_records = 0

with open(file_path, "r", encoding="latin1") as file:

    for line in file:

        total_records += 1

        # CDC positions are 1-based.
        # Python positions are 0-based.
        year = line[8:12].strip()
        month = line[12:14].strip()

        if year == "2024" and month in [f"{i:02d}" for i in range(1, 13)]:

            monthly_counts[month] += 1
            valid_records += 1

print("Total records:", total_records)
print("Valid 2024 records:", valid_records)
print("Monthly counts:", monthly_counts)
```

```python
month_names = {
    "01": "January",
    "02": "February",
    "03": "March",
    "04": "April",
    "05": "May",
    "06": "June",
    "07": "July",
    "08": "August",
    "09": "September",
    "10": "October",
    "11": "November",
    "12": "December"
}

monthly_data = pd.DataFrame({
    "month": list(month_names.keys()),
    "births": [monthly_counts[m] for m in month_names.keys()]
})

monthly_data["month_name"] = monthly_data["month"].map(month_names)

monthly_data
```

```python
from pathlib import Path

output_path = Path("data/monthly_births_2024.csv")

monthly_data.to_csv(output_path, index=False)

print("Saved:", output_path)
```
